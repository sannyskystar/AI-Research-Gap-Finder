```python
from html.parser import HTMLParser
import requests
import re

class TextExtractor(HTMLParser):
    def __init__(self):
        super().__init__()
        self.text = []

    def handle_data(self, data):
        text = data.strip()
        if text:
            self.text.append(text)

url = "https://arxiv.org/html/1809.00219"

html = requests.get(url).text

parser = TextExtractor()
parser.feed(html)

text = "\n".join(parser.text)

# Clean excessive whitespace
text = re.sub(r'\n+', '\n', text)
text = re.sub(r' +', ' ', text)

print("Extracted characters:", len(text))
print("\nFIRST 3000 CHARACTERS:\n")
print(text[:3000])
```

```python
from html.parser import HTMLParser

class ArticleTextExtractor(HTMLParser):
    def __init__(self):
        super().__init__()
        self.inside_article = False
        self.text = []

    def handle_starttag(self, tag, attrs):
        if tag == "article":
            self.inside_article = True

    def handle_endtag(self, tag):
        if tag == "article":
            self.inside_article = False

    def handle_data(self, data):
        if self.inside_article:
            data = data.strip()
            if data:
                self.text.append(data)

parser = ArticleTextExtractor()
parser.feed(html)

paper_text = "\n".join(parser.text)

print("Paper characters:", len(paper_text))
print(paper_text[:3000])
```

```python
import re

def extract_section(text, start_name, end_names):
    pattern = rf"\b{re.escape(start_name)}\b"
    match = re.search(pattern, text, re.IGNORECASE)

    if not match:
        return None

    start = match.end()

    end_positions = []
    for end_name in end_names:
        m = re.search(
            rf"\b{re.escape(end_name)}\b",
            text[start:],
            re.IGNORECASE
        )
        if m:
            end_positions.append(start + m.start())

    end = min(end_positions) if end_positions else len(text)

    return text[start:end].strip()


for name, text in papers.items():
    conclusion = extract_section(
        text,
        "Conclusion",
        ["References", "6", "7", "Acknowledgments"]
    )

    print("\n" + "="*70)
    print(name)
    print(conclusion[:3000] if conclusion else "Not found")
```

```python
def extract_numbered_section(text, section_number, next_section_number):
    pattern = rf'\b{re.escape(section_number)}\b\s*\n?\s*([A-Z][^\n]{{2,80}})'
    match = re.search(pattern, text)

    if not match:
        return None

    start = match.start()

    next_match = re.search(
        rf'\b{re.escape(next_section_number)}\b\s*\n?\s*[A-Z]',
        text[start + 20:]
    )

    end = start + 20 + next_match.start() if next_match else len(text)

    return text[start:end].strip()


for name, text in papers.items():
    limitations = extract_numbered_section(text, "4.4", "5")

    print("\n" + "="*70)
    print(name)
    print("\n", limitations[:3000] if limitations else "Not found")
```

```python
def extract_between(text, start, end=None):
    start_pos = text.lower().find(start.lower())

    if start_pos == -1:
        return ""

    start_pos += len(start)

    if end:
        end_pos = text.lower().find(end.lower(), start_pos)
        if end_pos != -1:
            return text[start_pos:end_pos].strip()

    return text[start_pos:].strip()


auto_evidence = {}

for name, text in papers.items():

    auto_evidence[name] = {
        "abstract": extract_between(text, "Abstract", "1 Introduction")[:5000],
        "introduction": extract_between(text, "1 Introduction", "2 Related Work")[:5000],
        "conclusion": extract_between(text, "5 Conclusion", "Acknowledgement")[:5000]
    }

    print(name)
    print("Abstract:", len(auto_evidence[name]["abstract"]))
    print("Introduction:", len(auto_evidence[name]["introduction"]))
    print("Conclusion:", len(auto_evidence[name]["conclusion"]))
```

```python
def generate_detailed_report(gaps, papers):
    report = "RESEARCH GAP REPORT\n"
    report += "=" * 60 + "\n\n"

    for i, (_, item) in enumerate(gaps.items(), 1):
        report += f"{i}. {item['gap']}\n"
        report += f"   Evidence: {', '.join(item['evidence'])}\n"

        for paper in item["evidence"]:
            report += f"   - {paper}: {papers[paper]['limitation']}\n"

        report += "\n"

    return report


detailed_report = generate_detailed_report(
    gap_engine,
    research_evidence
)

print(detailed_report)
```

```python
def analyze_papers(urls):
    urls = [u.strip() for u in urls.splitlines() if u.strip()]

    if not urls:
        return "Please enter at least one arXiv URL."

    paper_data = []

    for url in urls:
        paper_id = url.rstrip("/").split("/")[-1].replace("abs/", "")
        r = requests.get(f"https://arxiv.org/html/{paper_id}")

        if r.status_code != 200:
            continue

        parser = ArticleTextExtractor()
        parser.feed(r.text)

        text = re.sub(r"\s+", " ", " ".join(parser.text)).strip()
        lower = text.lower()

        paper_data.append({
            "id": paper_id,
            "artifacts": any(k in lower for k in
                             ["artifact", "artifacts", "distortion", "overshoot", "twisted"]),
            "generalization": any(k in lower for k in
                                  ["out-of-distribution", "unknown", "unseen"]),
            "real_world": "real-world" in lower,
            "data": any(k in lower for k in
                        ["dataset", "synthetic", "benchmark"])
        })

    # Shared evidence
    artifact_count = sum(p["artifacts"] for p in paper_data)
    generalization_count = sum(p["generalization"] for p in paper_data)
    data_count = sum(p["data"] for p in paper_data)

    report = "🔬 RESEARCH GAP REPORT\n"
    report += "=" * 55 + "\n\n"

    if artifact_count >= 2:
        report += "1. ARTIFACT / DISTORTION GAP\n"
        report += "Multiple papers discuss artifacts or distortion.\n"
        report += f"Evidence: {artifact_count}/{len(paper_data)} papers\n\n"

    if generalization_count >= 1:
        report += "2. GENERALIZATION GAP\n"
        report += "Unknown or out-of-distribution degradations remain a concern.\n"
        report += f"Evidence: {generalization_count}/{len(paper_data)} papers\n\n"

    if data_count >= 2:
        report += "3. DATA / BENCHMARK GAP\n"
        report += "The approaches rely on specific datasets or synthetic/benchmark data.\n"
        report += f"Evidence: {data_count}/{len(paper_data)} papers\n\n"

    report += "⚠️ These are candidate gaps based on textual evidence,\n"
    report += "not confirmed novel research gaps."

    return report


demo = gr.Interface(
    fn=analyze_papers,
    inputs=gr.Textbox(
        label="arXiv Paper URLs",
        placeholder="One URL per line"
    ),
    outputs=gr.Textbox(label="Research Gap Report", lines=20),
    title="AI Research Gap Finder",
    description="Compare research papers and identify candidate research gaps."
)

demo.launch(share=True)
```
